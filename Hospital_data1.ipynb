```python
import pandas as pd
import numpy as np
```

## Load the Data

```python
df=pd.read_csv("D:\\SQL_Classes\\Hospital_Data.csv")
df.head()
```

## Information of All columns

```python
df.info()
```

##  Convert Object to DateTime

```python
df["Admission_date"]=pd.to_datetime(df["Admission_date"],errors='coerce')
df["Discharge_date"]=pd.to_datetime(df["Discharge_date"],errors='coerce')
```

```python
print(df.dtypes)
```


## Check for Missing or Invalid Values

```python
print(df.isnull().sum())
```

##  Total Rows and Columns

```python
print(df.shape)  
```

##  Remove rows with missing dates

```python
df=df.dropna(subset=["Admission_date","Discharge_date"])
```

##  Check Again

```python
df.shape
```

```python
df.info()
```

# Export to CSV


```python
df.to_csv('Hospital_data.csv', index=False)  # index=False avoids writing row numbers (indices) to the file
```
